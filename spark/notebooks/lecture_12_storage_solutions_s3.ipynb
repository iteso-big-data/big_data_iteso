# Install AWS CLI

```python
!apt-get update
!apt-get install -y curl unzip vim
!curl -sSL "https://awscli.amazonaws.com/awscli-exe-linux-$(uname -m).zip" -o /tmp/awscliv2.zip
!unzip -q /tmp/awscliv2.zip -d /tmp
!/tmp/aws/install
!rm -rf /tmp/aws /tmp/awscliv2.zip
!mkdir ~/.aws
```

## Verify AWS CLI installation


```python
!aws --version
```

## List your bucket

```python
!aws s3 ls s3://bucket/path/ 
```

## Most common commands

```python
!aws s3 ls                                  
!aws s3 ls s3://bucket/path/               
!aws s3 ls s3://bucket/path/ --recursive --human-readable --summarize
```

### View a file

```python
!aws s3 cp s3://bucket/raw/events/file.json - | head
```

### Data Movement \& File Ops

```python
!aws s3 cp local_file s3://bucket/raw/      # upload
!aws s3 cp s3://bucket/raw/file.json ./     # download
!aws s3 sync ./data s3://bucket/raw/data/   # sync a directory
!aws s3 cp s3://bucket/src/file s3://bucket/dst/file
!aws s3 mv s3://bucket/src/file s3://bucket/dst/file
!aws s3 rm s3://bucket/path/file
!aws s3 rm s3://bucket/path/ --recursive    # delete a prefix
!aws s3 rb s3://bucket --force              # delete bucket
```
